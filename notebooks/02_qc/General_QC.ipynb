# Quality Control for Phosphoproteomics Data

Given your experimental setup (TMT-LC-MS/MS and LFQ-LC-MS/MS, multiple cell lines, time-course, stimulation conditions), here is a structured overview of the QC you should perform, organized by stage.

---

## 1. Raw Data / Peptide Detection Quality

These checks happen at the level of what was detected by the mass spectrometer.

### Essential

- **Missing value analysis** — what % of peptides are missing per sample? A sample with >30–40% missing values is suspicious and may need to be flagged or dropped.
- **Number of identified peptides/proteins per sample** — large drops in one replicate indicate technical problems (bad injection, failed enrichment, etc.)
- **Phosphorylation site localization score** — most pipelines (MaxQuant, Spectronaut, etc.) give a localization probability. You should filter to keep only confidently localized sites (typically >0.75 or Class I sites ≥0.75 in PhosphoRS/Andromeda scoring).
- **Replicate count per peptide** (`n_rep`) — peptides detected in only 1 replicate carry much less confidence.

### Recommended

- **Peptide charge state and mass error distributions** — should be consistent across samples; outliers suggest instrument drift.
- **Sequence coverage per protein** — not critical for phosphoproteomics but good to track.

---

## 2. Quantification Quality (Sample-level)

These check whether your samples are comparable to each other.

### Essential

- **Total intensity / signal distribution per sample** — box plots of raw intensities. All samples should have similar medians before normalization. Large deviations point to pipetting errors or enrichment efficiency differences.
- **Coefficient of Variation (CV) between replicates** — for the same condition and time point, replicates should agree. CVs >20–30% in many peptides is a warning sign.
- **Correlation matrix / heatmap between samples** — replicates should cluster together. If a replicate clusters with a different condition, something went wrong.

### Recommended

- **PCA (Principal Component Analysis)** — this is one of the most informative QC plots. PC1/PC2 should separate conditions or time points in a biologically meaningful way. Replicates should group tightly. *Outlier replicates become immediately obvious.*
- **Sample-to-sample distance heatmap** — complementary to PCA.

---

## 3. Normalization QC

You need to verify that your normalization worked as intended.

### Essential

- **Before/after normalization box plots** — medians should align after normalization.
- **CV before vs after normalization** — normalization should reduce CV between replicates without distorting biological signal.
- **Check that the starve control is behaving as expected** — since your fold change (FC) is calculated relative to starve, the starve samples should have FC ≈ 0 (in log2 space) and low variance.

### Recommended

- **MA plots** (log ratio vs mean average) between pairs of replicates — random scatter around 0 is expected; systematic trends indicate normalization issues.

---

## 4. Biological / Experimental QC

These verify that your experiment worked biologically.

### Essential

- **Known pathway marker behavior** — you are working on MAPK/ERK, so peptides like ERK1/2 pT202/pY204, RSK pS380, MEK pS217/S221 should show the expected rapid increase after EGF stimulation. If they don't, the experiment failed.
- **Full vs. starve comparison** — starved samples should show globally reduced phosphorylation compared to full media. A global reduction in signal intensity (or number of detected phosphopeptides) is expected.
- **Time course monotonicity check for known sites** — for EGF-responsive sites, the profile should peak early (1–5 min) and decay. Insulin-responsive sites (AKT pathway) tend to peak later and sustain longer.

### Recommended

- **Enrichment analysis on detected proteins** — are kinases and signaling proteins enriched in your phosphoproteome as expected? GO/KEGG enrichment.
- **Overlap between replicates (Venn diagrams)** — what fraction of peptides is consistently detected across all replicates?
- **TMT batch effect check** *(for your TMT data specifically)* — if your TMT experiments were run in multiple batches, you need to check for and correct batch effects before pooling.

---

## 5. Specific to Your Experimental Design

### For TMT vs LFQ comparison (WT vs mutants)

- These two datasets **cannot be directly compared quantitatively** without careful harmonization. The QC should verify that the dynamic range and detection overlap are sufficient for your classifier approach.
- **Peptide overlap between TMT and LFQ datasets** — how many phosphopeptides are shared? Only shared peptides can be used to transfer cluster labels.

### For the mutant cell lines

- Verify the mutation was introduced correctly by checking that the mutated peptide (Ala substitution) is detected and the original phosphopeptide at that site is absent.

---

## 6. Priority Order

| Priority | QC Step |
|:---:|---|
| 1 | Missing value analysis |
| 2 | Intensity distribution box plots (raw + normalized) |
| 3 | CV between replicates |
| 4 | PCA |
| 5 | Correlation heatmap |
| 6 | Known EGF/MAPK marker behavior |
| 7 | Localization score filtering |
| 8 | TMT batch effect (if multiple batches) |
